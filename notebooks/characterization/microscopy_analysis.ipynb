# Electron Microscopy Image Analysis (SEM / TEM / HRTEM)

This notebook demonstrates quantitative microscopy analysis:
- Mathematical physical scale bar rendering with true physical units
- Otsu segmentation and morphological filtering of nanoparticles
- Equivalent circular diameter statistics ($d_{10}, d_{50}, d_{90}$)
- Log-normal particle size distribution fitting
- HRTEM 2D FFT power spectrum and atomic lattice fringe $d$-spacing profiling.

```python
import sys
from pathlib import Path
root_path = Path.cwd().parent.parent if Path.cwd().name == 'characterization' else Path.cwd()
if str(root_path) not in sys.path:
    sys.path.insert(0, str(root_path))

import matplotlib.pyplot as plt
import numpy as np
from parsers.characterization import parse_microscopy_image
from modules.characterization.microscopy import (
    segment_particles,
    analyze_hrtem_lattice_fringes,
    plot_microscopy_with_scalebar,
    plot_segmentation_overlay,
    plot_particle_size_distribution,
    plot_hrtem_fft_panel,
)
from scripts.utils import apply_publication_style

apply_publication_style()
print('Microscopy analysis modules loaded.')
```

## 1. SEM Nanoparticle Segmentation & Size Distribution
Load FESEM catalyst image with physical calibration of 0.50 nm/pixel.

```python
sem_path = root_path / 'data' / 'examples' / 'characterization' / 'microscopy' / 'sem_nanoparticles.png'
sem_data = parse_microscopy_image(sem_path, pixel_size_nm=0.50)
seg_res = segment_particles(sem_data, min_area_px=15, clear_borders=True)
stats = seg_res['statistics']
print(f'Segmented {stats["particle_count"]} particles')
print(f'Mean Diameter: {stats["mean_diameter_nm"]:.1f} +/- {stats["std_diameter_nm"]:.1f} nm')
print(f'Median Diameter (d50): {stats["median_diameter_nm"]:.1f} nm (d10={stats["p10_diameter_nm"]:.1f}, d90={stats["p90_diameter_nm"]:.1f} nm)')

fig1, ax1 = plot_segmentation_overlay(sem_data, seg_res, title='FESEM Catalyst Nanoparticles & Segmentation')
plt.show()

fig2, ax2 = plot_particle_size_distribution(seg_res, bins=15, title='Nanoparticle Size Distribution')
plt.show()
```

## 2. HRTEM 2D FFT Lattice Fringe Analysis
Load high-resolution TEM crop calibrated at 0.018 nm/pixel (0.18 A/pixel).

```python
hrtem_path = root_path / 'data' / 'examples' / 'characterization' / 'microscopy' / 'hrtem_lattice_crop.png'
hrtem_data = parse_microscopy_image(hrtem_path, pixel_size_nm=0.018)
fringe_res = analyze_hrtem_lattice_fringes(hrtem_data, fringe_direction='horizontal')
print(f'Lattice fringe d-spacing: {fringe_res["d_spacing_angstrom"]:.2f} A ({fringe_res["d_spacing_nm"]:.3f} nm)')

fig, axes = plot_hrtem_fft_panel(hrtem_data, fringe_res, title=f'HRTEM Atomic Lattice Resolution (d = {fringe_res["d_spacing_angstrom"]:.2f} A)')
plt.show()
```
